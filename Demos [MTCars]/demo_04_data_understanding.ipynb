## Why is my output so unreadable?


```python
# Create the Factor Variable
# We map the numeric values to descriptive labels and set the type to 'category'
df['vs_cat'] = df['vs'].map({0: 'V-shaped', 1: 'Straight'}).astype('category')
```

```python
# This makes the output much more intuitive
print("\nAverage Horsepower (hp) by Engine Type:")
print(df.groupby("vs_cat")["hp"].mean())
```

## We created a categorical variable instead of a repeating STRING above. Why?

## Lets create a string instead and compare

```python
# Create the new String variable vs_str
# This preserves the original 'vs' column while adding a descriptive one
df['vs_str'] = df['vs'].map({0: 'V-shaped', 1: 'Straight'})

# Inspect the Data
# You can now see both the encoded value and the human-readable string
print("New DataFrame view with vs_str:")
print(df[['car_name', 'vs', 'vs_cat','vs_str', 'hp']].head())
```

```python
# Compute Row Size (Memory Usage)
# We use deep=True to ensure we measure the actual content of the strings
mem = df[['vs','vs_cat', 'vs_str']].memory_usage(deep=True)

print("--- Memory Usage Comparison ---")
print(f"Memory for 'vs' (Integer):      {mem['vs']} bytes")
print(f"Memory for 'vs_cat' (Categorical):      {mem['vs_cat']} bytes")
print(f"Memory for 'vs_str' (String):   {mem['vs_str']} bytes")

# Calculate per-row average
row_count = len(df)
print(f"\nAverage size per row (Integer): {mem['vs'] / row_count:.2f} bytes")
print(f"\nAverage size per row (Categorical): {mem['vs_cat'] / row_count:.2f} bytes")
print(f"Average size per row (String):  {mem['vs_str'] / row_count:.2f} bytes")
```

## The Difference: 
You will notice that the vs column (integers) takes about 8 bytes per row, while the vs_str column (strings) takes roughly 65 bytes per row. The categorical takes a little less.

Why? Integers are stored as fixed-width binary values. Strings are far more complex; they require memory for the characters themselves plus "overhead" (pointers) that tell Python where that string starts and ends in memory.

The Impact: While this doesn't matter for 6 rows, if you were analyzing a dataset of 10 million car registrations, the string version would take ~650MB while the integer version would only take ~80MB. This is why we often perform our "heavy lifting" (math and filtering) on numeric codes and only use strings for final display.

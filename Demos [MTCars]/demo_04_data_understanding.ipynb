Import Data

```python
import pandas as pd

# Extended MTCARS Dataset (partial)
data = {
    "car_name": ["Mazda RX4", "Mazda RX4", "Datsun 710", "Datsun 710", "Hornet Sportabout", "Hornet Sportabout"],
    "mpg": [21.0, 21.4, 22.8, 23.5, 18.7, 18.1],
    "cyl": [6, 6, 44, 4, 8, 8],
    "disp": [160.0, 160.0, 108.0, 108.0, 360.0, 360.0],
    "hp": [110, 110, 93, 95, 175, 180],
    "drat": [3.90, 3.90, 3.85, 3.85, 3.15, 3.15],
    "wt": [2.620, 2.625, 2.320, 2.315, 3.440, 3.460],
    "qsec": [16.46, 16.44, 18.61, 18.50, 1702, 17.05],
    "vs": [0, 0, 1, 1, 0, 0],
    "am": [1, 1, 1, 1, 0, 0],
    "gear": [4, 4, 4, 4, 3, 3],
    "carb": [4, 4, 1, 1, 2, 2],
    "car_id": [101, 101, 102, 102, 103, 103],
    "year": [2000, 2002, 2000, 2002, 2000, 2002]
}

# Convert the dataset into a Pandas DataFrame
df = pd.DataFrame(data)

print(df.head())
```

```python
# Average MPG by Year (Is fuel efficiency improving over time?)
df.groupby("year")["mpg"].mean()

```

```python
# Average Horsepower (hp) by Engine Type (vs: 0 = V-shaped, 1 = straight)
df.groupby("vs")["hp"].mean()
```

## Why is my output so unreadable?


Create the Factor Variable
# We map the numeric values to descriptive labels and set the type to 'category'
df['vs'] = df['vs'].map({0: 'V-shaped', 1: 'Straight'}).astype('category')

```python
# This makes the output much more intuitive
print("\nAverage Horsepower (hp) by Engine Type:")
print(df.groupby("vs")["hp"].mean())
```

## We created a categorical variable instead of a repeating STRING above. Why?

## Lets create a string instead and compare

```python
# Create the new String variable vs_str
# This preserves the original 'vs' column while adding a descriptive one
df['vs_str'] = df['vs'].map({0: 'V-shaped', 1: 'Straight'})

# Inspect the Data
# You can now see both the encoded value and the human-readable string
print("New DataFrame view with vs_str:")
print(df[['car_name', 'vs', 'vs_str', 'hp']].head())
```
